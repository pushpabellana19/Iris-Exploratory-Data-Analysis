```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Exploratory Data Analysis: Iris Dataset

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Species Count
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="species")
plt.title("1. Count of Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()

# 2. Sepal Length Distribution
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x="sepal_length", hue="species",
             kde=True, multiple="layer")
plt.title("2. Distribution of Sepal Length")
plt.show()

# 3. Petal Length by Species
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="species", y="petal_length")
plt.title("3. Petal Length by Species")
plt.show()

# 4. Sepal Length vs Sepal Width
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="sepal_length", y="sepal_width",
                hue="species", s=70)
plt.title("4. Sepal Length vs Sepal Width")
plt.show()

# 5. Correlation Heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(df.select_dtypes(include="number").corr(),
            annot=True, cmap="coolwarm", fmt=".2f")
plt.title("5. Correlation Between Iris Measurements")
plt.show()

print("All 5 visualizations created successfully!")
```

Exploratory Data Analysis-Iris Dataset
1. Observations from Visualizations

Visualization 1: Count of Iris Species
The dataset contains 150 flower samples, with 50 samples from each species: Setosa, Versicolor, and Virginica.

Visualization 2: Distribution of Sepal Length
Sepal lengths vary across the three species. Their distributions overlap, so sepal length alone does not perfectly distinguish every species.

Visualization 3: Petal Length by Species
Setosa generally has the shortest petals, while Virginica generally has the longest. Petal length helps distinguish the species.

Visualization 4: Sepal Length vs Sepal Width
The scatter plot shows patterns among the species. Setosa tends to form a more distinct group, while Versicolor and Virginica overlap more.

Visualization 5: Correlation Heatmap
Petal length and petal width have a strong positive correlation. Flowers with longer petals generally also have wider petals.

2.Top 3 Insights

1. Petal measurements are useful for classification: Petal length and width help distinguish Iris species.
2. Setosa is easier to distinguish: Its petal measurements are generally smaller than those of the other two species.
3. Some species overlap: Versicolor and Virginica share some measurement ranges, making them harder to separate using a single feature.

 Conclusion :

Exploratory Data Analysis helps us understand the Iris dataset through statistical summaries and visualizations. The analysis shows that petal measurements are useful for identifying differences between species. Visualizations make patterns and relationships easier to understand.
